# Mapas

```python
# 1. folium - la mas popular y facil
import folium 

# mapa basico
m = folium.Map(location=[4.6097, -74.0817], zoom_start=10) # bogota
m.save('mapa.html')
m # en jupyter se muestra directamente
       
```

```python
# 2. Plotly - Mapas interactivos avanzados

import plotly.graph_objects as go
import plotly.express as px

# Mapa con marcadores
fig = go.Figure(go.Scattermapbox(
    lat=[4.6097],
    lon=[-74.0817],
    mode='markers',
    marker=dict(size=14, color='red'),
    text=['Bogotá'],
))

fig.update_layout(
    mapbox_style="open-street-map",
    mapbox=dict(center=dict(lat=4.6097, lon=-74.0817), zoom=10),
    margin={"r":0,"t":0,"l":0,"b":0}
)
fig.show()
```

```python
import folium

# Mapa básico de Colombia
m = folium.Map(location=[4.6, -74.08], zoom_start=6)

# Agregar marcadores de ciudades principales
ciudades = {
    'Bogotá': [4.6097, -74.0817],
    'Medellín': [6.2442, -75.5812],
    'Cali': [3.4516, -76.5320],
    'Barranquilla': [10.9685, -74.7813],
    'Cartagena': [10.3910, -75.4794],
    'Bucaramanga': [7.1254, -73.1198]  # Coordenadas de Bucaramanga
}

# Direcciones específicas para mostrar en el popup
direcciones = {
    'Bogotá': 'Bogotá, Cundinamarca, Colombia',
    'Medellín': 'Medellín, Antioquia, Colombia',
    'Cali': 'Cali, Valle del Cauca, Colombia',
    'Barranquilla': 'Barranquilla, Atlántico, Colombia',
    'Cartagena': 'Cartagena de Indias, Bolívar, Colombia',
    'Bucaramanga': 'Bucaramanga, Santander, Colombia'
}

for ciudad, coords in ciudades.items():
    folium.Marker(
        coords,
        popup=f"{ciudad} - {direcciones[ciudad]} (Lat: {coords[0]}, Lon: {coords[1]})",
        tooltip=ciudad,
        icon=folium.Icon(color='red', icon='info-sign')
    ).add_to(m)

# Mostrar el mapa

m
```
